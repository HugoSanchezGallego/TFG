# 08 · Subida del modelo final v4 a Hugging Face

**Objetivo:** publicar el modelo v4 (detector final de la fase 2) en Hugging Face Hub para garantizar su reproducibilidad y disponibilidad.

**Contexto:** el modelo v4, entrenado a 1280px, ha resultado ser el mejor detector tras la comparativa sobre test. Se publica en Hugging Face como modelo de referencia del sistema, acompañado de un README con sus métricas.

````python
import os
from huggingface_hub import HfApi, upload_file
from pathlib import Path

REPO_ROOT = Path("../..").resolve()
MODEL_PATH = REPO_ROOT / "models" / "best_v4.pt"

# El token ya está guardado como variable de entorno desde el notebook de setup
token = os.environ["HUGGINGFACE_TOKEN"]
api = HfApi(token=token)
info = api.whoami()

REPO_ID = f"{info['name']}/futbol-detector-v4"
print(f"Usuario: {info['name']}")
print(f"Repositorio destino: {REPO_ID}")
print(f"Modelo: {MODEL_PATH} ({MODEL_PATH.stat().st_size / (1024*1024):.2f} MB)")
````

````python
# Crear el repositorio si no existe
api.create_repo(repo_id=REPO_ID, repo_type="model", exist_ok=True, private=False)
print(f"Repositorio listo: https://huggingface.co/{REPO_ID}")
````

````python
# Subir el modelo
url = upload_file(
    path_or_fileobj=str(MODEL_PATH),
    path_in_repo="best.pt",
    repo_id=REPO_ID,
    token=token,
    commit_message="Modelo v4 (1280px) - detector final de la fase 2"
)
print(f"Modelo subido correctamente")
print(f"URL: {url}")
````

````python
readme_content = f"""---
license: mit
tags:
- yolov8
- object-detection
- football
- soccer
- sports-analytics
---

# Futbol Detector v4 (1280px)

Modelo YOLOv8s entrenado a resolución 1280px para detectar jugadores, árbitros y balón en vídeos de fútbol profesional. Es el detector final del Trabajo de Fin de Grado tras una comparativa de cuatro modelos.

## Contexto

Forma parte del TFG *"Diseño de un sistema de asistencia al análisis táctico deportivo basado en procesamiento de imagen"* (Grado en Ingeniería Informática, Universidad Alfonso X el Sabio).

## Evolución de los modelos

El modelo es el resultado de un proceso iterativo de cuatro versiones, comparadas sobre un conjunto de test independiente:

| Modelo | Descripción | Balón | Jugador | Árbitro | mAP50 | FPS |
|---|---|---|---|---|---|---|
| v1 | 152 img, 640px | 0.398 | 0.893 | 0.143 | 0.478 | 79 |
| v2 | 743 img, 640px | 0.445 | 0.949 | 0.516 | 0.637 | 84 |
| v3 | 743 img + augmentation, 640px | 0.466 | 0.929 | 0.566 | 0.654 | 57 |
| **v4** | **743 img, 1280px** | **0.574** | **0.962** | **0.581** | **0.706** | 31 |

El v4 ofrece la mejor precisión en todas las clases, destacando la mejora en la detección del balón gracias a la mayor resolución.

## Clases

- ball (balón)
- player (jugador)
- referee (árbitro)

## Uso

```python
from ultralytics import YOLO
from huggingface_hub import hf_hub_download

model_path = hf_hub_download(repo_id="{REPO_ID}", filename="best.pt")
model = YOLO(model_path)

results = model("ruta/al/video.mp4", imgsz=1280)
```

## Repositorio del proyecto

[https://github.com/HugoSanchezGallego/TFG](https://github.com/HugoSanchezGallego/TFG)
"""

api.upload_file(
    path_or_fileobj=readme_content.encode(),
    path_in_repo="README.md",
    repo_id=REPO_ID,
    commit_message="README con comparativa de modelos"
)

print(f"README subido")
print(f"Página del modelo: https://huggingface.co/{REPO_ID}")
````
